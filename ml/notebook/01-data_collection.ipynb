# 01 — Data Collection

This notebook synchronizes the latest KoboToolbox XLS export into the local `ml/data/raw` and `ml/data/archive` folders.

```python
from blended_learning.config.settings import settings
from blended_learning.ingestion.kobo_loader import KoboSync

kobo_cfg = settings.data_collection["kobo"]
output_cfg = kobo_cfg["output"]

print(f"Config file: {settings.config_path}")
print(f"Raw folder:   {settings.root / output_cfg['raw_dir']}")
print(f"Archive:      {settings.root / output_cfg['archive_dir']}")
print(f"Latest XLSX:  {output_cfg['latest_xlsx']}")
print(f"Latest file:  {output_cfg['latest_parquet']}")
```

## Run Kobo sync

This cell requires the following environment variables: `KOBO_BASE_URL`, `KOBO_TOKEN`, and `KOBO_ASSET_UID`.

```python
sync = KoboSync()
df_raw = sync.run()

print(f"Rows fetched: {len(df_raw):,}")
print(f"Columns fetched: {df_raw.shape[1]:,}")
df_raw.head()
```
